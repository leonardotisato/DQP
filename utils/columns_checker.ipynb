```python
import os
import pandas as pd
import re
```

```python
df = pd.read_csv("../Comune-di-Milano-Servizi-alla-persona-parrucchieri-estetisti(in).csv", sep=';',encoding='unicode_escape')

output_folder = "col_values"
os.makedirs(output_folder, exist_ok=True)

def clean_colname(name):
    name = name.strip().lower()
    name = re.sub(r'[^a-z0-9]+', '_', name)
    return name.strip('_')


for col in df.columns:
    cleaned = clean_colname(col)
    values = (
        df[col]
        .drop_duplicates()
    )
    values = sorted(values, key=lambda x: str(x))
    out_path = os.path.join(output_folder, f"{cleaned}.csv")
    pd.DataFrame(values, columns=[col]).to_csv(out_path, index=False)

    print(f"Saved: {out_path}")

print("Completato!")
```
